# Filtrado temático

Dado que la base de datos inicial contiene la totalidad de disposiciones publicadas en el BOE durante el período analizado, la validación exhaustiva de accesibilidad de los enlaces HTML y PDF se realizó únicamente sobre el subconjunto de registros previamente identificado como potencialmente relacionado con proyectos de energías renovables. Esta decisión permitió optimizar los recursos computacionales sin afectar a la calidad del conjunto de datos final objeto de estudio.

```python
# Crear una versión normalizada del título. Se aborda mayúsculas/miúsculas, tildes y caracteres especiales.
df_boe_23["titulo_norm"] = (
    df_boe_23["titulo"]
    .str.lower()
    .str.normalize("NFKD")  # Descompone caracteres acentuados (á → a + acento, ñ → n + tilde, etc.)
    .str.encode("ascii", errors="ignore")  # Convierte a ASCII eliminando los caracteres no representables (acentos, diéresis, etc.)
    .str.decode("utf-8")  # Convierte de nuevo los bytes ASCII a texto UTF-8
)

df_boe_23
```

```python
keywords_energy = [
    "parque eolico",
    "parque solar",
    "planta fotovoltaica",
    "instalacion fotovoltaica",
    "instalacion solar",
    "energia eolica",
    "energia solar",
    "subestacion electrica",
    "linea electrica",
    "linea de evacuacion",
    "evacuacion de energia",
    "alta tension",
    "muy alta tension",
    "generacion electrica"
]

keywords_permitting = [
    "autorizacion administrativa previa",
    "autorizacion administrativa de construccion",
    "declaracion de impacto ambiental",
    "evaluacion de impacto ambiental",
    "informacion publica",
    "utilidad publica",
    "ocupacion de bienes",
    "expropiacion forzosa"
]

# Unifica todas las palabras clave eliminando duplicados
keywords = sorted(set(keywords_energy + keywords_permitting))

# Construye la expresión regular
regex = "|".join(keywords)

# Filtra los registros candidatos
df_energia_23 = df_boe_23[
    df_boe_23["titulo_norm"].str.contains(regex, na=False)
]
```
